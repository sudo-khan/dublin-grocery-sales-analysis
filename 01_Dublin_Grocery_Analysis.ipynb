```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create our grocery sales dataset
data = {
    'Transaction_ID': ['TXN-1001', 'TXN-1002', 'TXN-1003', 'TXN-1004', 'TXN-1005', 'TXN-1006', 'TXN-1007', 'TXN-1008'],
    'Date': ['2026-09-01', '2026-09-01', '2026-09-02', '2026-09-02', '2026-09-03', '2026-09-03', '2026-09-04', '2026-09-04'],
    'Category': ['Meat / Poultry', 'Dairy', 'Grains / Spices', 'Meat / Poultry', 'Grains / Spices', 'Dairy', 'Produce', 'Meat / Poultry'],
    'Product': ['Halal Chicken Fillet (1kg)', 'Greek Style Yogurt (500g)', 'Premium Basmati Rice (5kg)', 'Halal Lamb Chops (1kg)', 'Wholemeal Atta Flour (10kg)', 'Whole Milk (2 Litre)', 'Fresh Tomatoes (1kg bag)', 'Halal Chicken Fillet (1kg)'],
    'Qty_Sold': [45, 30, 12, 15, 20, 60, 40, 55],
    'Unit_Cost': [5.50, 1.10, 6.00, 9.00, 4.50, 0.95, 0.80, 5.50],
    'Retail_Price': [8.50, 2.20, 11.99, 15.50, 8.99, 1.85, 1.99, 8.50]
}

df = pd.DataFrame(data)

# Calculate financial metrics
df['Total_Revenue'] = df['Qty_Sold'] * df['Retail_Price']
df['Total_Cost'] = df['Qty_Sold'] * df['Unit_Cost']
df['Net_Profit'] = df['Total_Revenue'] - df['Total_Cost']

# Group by product to summarize total profit
summary = df.groupby('Product')['Net_Profit'].sum().reset_index().sort_values(by='Net_Profit', ascending=False)

print("--- Net Profit Summary (€) ---")
print(summary)

# Generate visual bar chart
plt.figure(figsize=(10, 4))
sns.barplot(data=summary, x='Net_Profit', y='Product', palette='crest')
plt.title('Top Profit Generating Products (€)')
plt.xlabel('Net Profit (€)')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define raw grocery sales data
data = {
    'Transaction ID': ['TXN-1001', 'TXN-1002', 'TXN-1003', 'TXN-1004', 'TXN-1005', 'TXN-1006', 'TXN-1007', 'TXN-1008'],
    'Date': ['2026-09-01', '2026-09-01', '2026-09-02', '2026-09-02', '2026-09-03', '2026-09-03', '2026-09-04', '2026-09-04'],
    'Item Category': ['Meat / Poultry', 'Dairy', 'Grains / Spices', 'Meat / Poultry', 'Grains / Spices', 'Dairy', 'Produce', 'Meat / Poultry'],
    'Product Name': ['Halal Chicken Fillet (1kg)', 'Greek Style Yogurt (500g)', 'Premium Basmati Rice (5kg)', 'Halal Lamb Chops (1kg)', 'Wholemeal Atta Flour (10kg)', 'Whole Milk (2 Litre)', 'Fresh Tomatoes (1kg bag)', 'Halal Chicken Fillet (1kg)'],
    'Qty Sold': [45, 30, 12, 15, 20, 60, 40, 55],
    'Unit Cost (€)': [5.50, 1.10, 6.00, 9.00, 4.50, 0.95, 0.80, 5.50],
    'Retail Price (€)': [8.50, 2.20, 11.99, 15.50, 8.99, 1.85, 1.99, 8.50]
}

df = pd.DataFrame(data)

# 2. Compute calculated financial fields
df['Total Revenue (€)'] = df['Qty Sold'] * df['Retail Price (€)']
df['Total Cost (€)'] = df['Qty Sold'] * df['Unit Cost (€)']
df['Net Profit (€)'] = df['Total Revenue (€)'] - df['Total Cost (€)']

# 3. Save directly as an Excel file in your current folder
file_path = 'Dublin_Grocery_Sales_Complete.xlsx'
df.to_excel(file_path, index=False)
print(f"File created successfully at: {file_path}")
```
